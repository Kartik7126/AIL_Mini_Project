# Credit-card fraud: EDA and anomaly detection

This notebook walks through class imbalance, amount/time structure, and the trained detectors in `frauddet`.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path('..').resolve() / 'src'))

import pandas as pd
import matplotlib.pyplot as plt
from frauddet.data import load_transactions, split_supervised
from frauddet.config import TARGET, METRICS_PATH

df = load_transactions()
df.head()
```

```python
print(df.shape)
print(df[TARGET].value_counts())
print('fraud rate', df[TARGET].mean())
df[['Time', 'Amount']].describe()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df[TARGET].value_counts().plot(kind='bar', ax=axes[0], color=['#3b82f6', '#ef4444'])
axes[0].set_title('Class counts')
df.groupby(TARGET)['Amount'].median().plot(kind='bar', ax=axes[1], color=['#3b82f6', '#ef4444'])
axes[1].set_title('Median amount by class')
plt.tight_layout()
```
